```
import numpy as np
class GeometricSieve:
 def shape_parameter(self, pos):
 a = np.linalg.norm(pos[1] - pos[0])
 b = np.linalg.norm(pos[2] - pos[1])
 c = np.linalg.norm(pos[0] - pos[2])
 s = (a+b+c)/2.0
 area_sq = max(s*(s-a)*(s-b)*(s-c),0.0)
 if area_sq < 1e-18: return 0.0
 area = np.sqrt(area_sq)
 r = area/(s+1e-18)
 R = (a*b*c)/(4.0*area+1e-18)
 return r/(R+1e-18)

s = GeometricSieve()
pos_3d = np.array([[1,0,0.2],[-0.5,0.866,1.4],[-0.5,-0.866,-0.7]])
print(f"3D varying z q = {s.shape_parameter(pos_3d):.4f} stable")
```
